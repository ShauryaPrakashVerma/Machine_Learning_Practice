# **Cross Validation**

### How can we fix overfitting with cross-validation?
```
Cross validation can't be used to fix overfitting.
We can only smell overfitting with it.
Large variation in cross-validation scores indicates that the model is unstable. One possible cause of this instability is overfitting, but it is not the only cause
```

````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
````

````python
from sklearn.model_selection import cross_val_score, KFold, train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.datasets import load_diabetes
from sklearn.metrics import mean_squared_error, r2_score
````

````python
dataset = load_diabetes()
X, y = dataset.data , dataset.target
````

````python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.30, random_state=1)
````

````python
model = LinearRegression()
````

````python
# TYPES OF CROSS VALIDATION

# kf = KFold(
#     n_splits=5,
#     shuffle=True,
#     random_state=42
# )



# StratifiedKFold(...)
# GroupKFold(...)
# TimeSeriesSplit(...)
# RepeatedKFold(...)
# LeaveOneOut(...)
````

````python
# scores = cross_val_score(model, X_train, y_train, cv =kf)
````

````python
scores = cross_val_score(model, X_train, y_train, cv =5)


# the above statement is equivalent to the below two statements
# kf = KFold(n_splits=5)
# scores = cross_val_score(model, X, y, cv=kf)
````

````python
scores
````

````python
scores.mean()
````

````python
model.fit(X_train,y_train)
````

````python
y_pred = model.predict(X_test)
````

````python
print("Predictions:")
print(y_pred)
````

````python
print("\nActual Values:")
print(y_test)
````

````python
print("\nMean Squared Error:", mean_squared_error(y_test, y_pred))
print("R² Score:", r2_score(y_test, y_pred))
````
