### Pandas Export
 - to_csv
 - to_excel
 - to_html
 - to_json
 - to_sql

```python
import pandas as pd
```

## to_csv

```python
df = pd.read_csv('batter.csv')
df
```

```python
# export a dataframe to a csv file
df.to_csv('batter_duplicate.csv')
```

## to_excel

```python
df.to_excel('ballebaaz.xlsx')
```

```python
df.to_excel('ballebaaji.xlsx', sheet_name="Sheet2")
```

```python
temp1 = df[df['runs'] > 1000]  # Or however you defined it
temp2 = df.copy()
```

```python
# to create an excel file with multiple sheets , or to work on multiple sheets
with pd.ExcelWriter('ballebaaz.xlsx') as writer:
  temp1.to_excel(writer, sheet_name = "Sheet 1")
  temp2.to_excel(writer, sheet_name = "Sheet 2")
```

## to_html

```python
df.pivot_table(index = 'batter', columns='runs').to_html('html_file.html')
```
