# 2. Numerical Data

### a. Histogram

```python
import matplotlib.pyplot as plt
plt.hist(df['Age'])
```

### b. Distplot

```python
# sns.displot(data=df, x='Age', kind='kde', fill='True')
sns.displot(data=df, x='Age', kde='True')


# can also give idea of skewness
```

### c. Boxplot

image.png

- IQR = Q3 - Q1
- minimum = Q1 - 1.5*IQR
- maximum = Q3 + 1.5*IQR

```python
# It gives you five number summary
# Can be used to identify outliers, in a very noisy dataset
# values less than minimum or greater than maximum are outliers
```

```python
sns.boxplot(df['Fare'])
# sns.boxplot(df['Age'])
```

```python
sns.boxplot(data=df, x='Fare')
```

```python
# Another way to calculate skewness
df['Age'].skew()

# 0 --> pefectly skewed
# 1 --> positively skewed
# -1 --> negatively skewed
```
